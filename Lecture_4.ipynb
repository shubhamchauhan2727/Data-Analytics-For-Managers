```python
#byjus case
import pandas as pd
import numpy as np
df=pd.read_excel("/content/Lecture2_data.xlsx")
```

```python
# Decision making structure
# Nested IF statement in python
revenue = df["Revenue_Lakh"].mean()
rating= df["Customer_Rating"].mean()

if revenue >=300:
  print("Revenue is high")

  if rating >=4:
    print("Customer satisfaction is also good")
  else:
      print("Customer satisfaction needs improvement")
else:
      print("revenue is low")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()

# Make a decision based on the rating
if rating >= 4.5:
  print("Excellent")

elif rating >= 4.0:
  print("good")

elif rating >= 3.0:
  print("Average")

else:
  print("needs improvement")
```

```python
#for loop
for rating in df["Customer_Rating"]:
 if rating>=4:
  print("good rating")
else:
  print("needs improvement")
```

```python
#for loop with range () iloc=integer location
for i in range(10):
  print(df["Product"].iloc[i],df["Revenue_Lakh"].iloc[i])
```

```python
#while loop in python
#while loop and i,j is variable to track
i=0
while i<5:
  revenue=df["Revenue_Lakh"].iloc[i]
  if revenue >=300:
    print("high revenue:",revenue)
  else:
      print("low revenue:",revenue)
  i=i+1
```

```python
import pandas as pd
#excel file path
FILE="/content/Starbucks_India_Sales_Dataset.xlsx"

#load the excel sheets
df=pd.read_excel(FILE,sheet_name="Sales_Data")
products_df=pd.read_excel(FILE,sheet_name="Products")
store_df=pd.read_excel(FILE,sheet_name="Stores")

#check the data
print("Sales table:",df.shape)
print("Products:",products_df.shape)
print("Stores:",store_df.shape)

#display first 5 rows
df.head()
```

```python
 #list: the starbucks menu(8 products)
menu=["Cappuccino","Latte","Americano","Cold Brew","Frappuccino","Masala Chai","Blueberry Muffin","Veg Sandwich"]

#tuple:one stores fixed details{store id ,city}
store=("S01","Connaught Place","Delhi")

#set:cities where starbucks sells (the duplicate 'delhi' is removed automatically)
cities={"Delhi","Mumbai","Bengaluru","Hydrabad","Pune","Delhi"}

#dictionary:products-> base price in rs
price={"Cappuccino":250,"Latte":270,"Americano":220,"Cold Brew":300,"Frappuccino":350,"Masala Chai":200,"Blueberry Muffin":180,"Veg Sandwich":240}

print("List:",menu)
print("Tuple:",store)
print("Set:",cities,"<-'Delhi'typed twice,stored once")
print("Dictionary:",price)
print()
print(type(menu),type(store),type(cities),type(price))
```

Section 2 : Assessing the elements of data structure


```python
# list
print("First item:",menu[0])
print("Last item:",menu[-1])
print("First three:",menu[0:3])
print("Every 2nd item",menu[::2])

#tuple
print("\nStore ID:",store[0])
print("Store city:"),store[2]

#dictionary
print("\nPrice of Latte:",price["Latte"])
print("Safe lookup:",price.get("Pizza","Not on menu"))

#set
print("\nIS Pune a Starbucks city?","Pune" in cities)
print("IS jaipur a Starbucks city?","Jaipur" in cities)
```

```python
revenue_list=df["Revenue"].tolist()
#buil-in functions on the revenue list(100 orders)
print("Number of orders:",len(revenue_list))
print("Total revenue:RS",round(sum(revenue_list),2))
print("Average order value:RS",round(sum(revenue_list)/len(revenue_list),2))
print("Biggest order:RS",max(revenue_list))
print("Smallest order:RS",min(revenue_list))
print("Top 5 orders:",sorted(revenue_list,reverse=True)[:5])
```
